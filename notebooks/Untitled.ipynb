```python
from lie_learn.representations.SO3 import spherical_harmonics
import lie_learn.spaces.S2 as S2
import numpy as np
```

```python
def generate_sh(L, theta, phi):
    irreps = np.arange(L + 1)
    ls = [[ls] * (2 * ls + 1) for ls in irreps]
    ls = np.array(
       [ll for sublist in ls for ll in sublist]
    )  # 0, 1, 1, 1, 2, 2, 2, 2, 2, ...
    
    ms = [list(range(-ls, ls + 1)) for ls in irreps]
    ms = np.array(
       [mm for sublist in ms for mm in sublist]
    )  # 0, -1, 0, 1, -2, -1, 0, 1, 2, ...

    Y = spherical_harmonics.sh(
       ls[:, None, None],
       ms[:, None, None],
       theta[None, :, :],
       phi[None, :, :],
       field="real",
       normalization="quantum",
       condon_shortley=True,
    )

    return Y
```

```python
theta, phi = S2.meshgrid(b=20, grid_type="Driscoll-Healy")
Y = generate_sh(1, theta, phi)
```
